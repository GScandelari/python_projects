# Control Flow in Python

Programs rarely run from top to bottom in a straight line. **Control flow** is the order in which individual statements and instructions are executed — and it is shaped by conditions and loops.

In this notebook you will learn:

- How to make decisions with `if`, `elif`, and `else`
- How to repeat actions with `while` loops
- How to iterate over sequences with `for` loops and `range()`
- How to use `break`, `continue`, and loop `else` clauses
- Common patterns such as accumulators, counters, and manual min/max

## 1. Conditional Statements

A conditional statement lets your program choose between different paths based on whether a condition is `True` or `False`.

### Syntax

```python
if condition:
    # runs when condition is True
elif another_condition:
    # runs when the first condition is False but this one is True
else:
    # runs when none of the above conditions are True
```

**Key rules:**
- Use `==` for comparison, not `=` (which is assignment).
- Python evaluates conditions top-to-bottom and stops at the first `True` branch.
- `elif` and `else` are optional; you can have as many `elif` clauses as needed.
- Indentation (4 spaces or 1 tab) defines the block — it is not optional.

````python
# --- Basic if / elif / else ---
score = 73

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(f"Score: {score}  ->  Grade: {grade}")

# --- Nested if ---
# Conditions can be placed inside other conditions.
temperature = 22
is_raining = False

if temperature > 18:
    if is_raining:
        print("Warm but rainy — bring an umbrella.")
    else:
        print("Nice weather — go outside!")
else:
    print("Too cold — stay in.")

# --- Ternary (conditional expression) ---
# A compact one-liner for simple true/false choices.
# Syntax: value_if_true  if  condition  else  value_if_false
age = 20
status = "adult" if age >= 18 else "minor"
print(f"Age {age} -> {status}")

# --- Combining conditions with and / or / not ---
x = 15
if x > 10 and x < 20:
    print(f"{x} is between 10 and 20")

is_weekend = True
if is_weekend or x > 50:
    print("Either it is the weekend or x is large.")

if not is_raining:
    print("It is not raining.")
````

## 2. The `while` Loop

A `while` loop repeats its body **as long as its condition remains `True`**.

```python
while condition:
    # body — runs repeatedly
```

**Important:** you must ensure the condition eventually becomes `False`, or use `break` to exit — otherwise the loop runs forever (an *infinite loop*).

### Control keywords inside loops

| Keyword | Effect |
|---------|--------|
| `break` | Immediately exits the loop, skipping any remaining iterations |
| `continue` | Skips the rest of the current iteration and goes back to the condition check |

### When to prefer `while` over `for`

Use `while` when you do **not** know in advance how many iterations are needed — for example, reading user input until a valid value is entered, or waiting for a condition to change.

````python
# --- Basic while: countdown ---
count = 5
while count > 0:
    print(count)
    count -= 1          # CRUCIAL: update the variable driving the condition
print("Go!")

print()

# --- while with break ---
# Search for the first number divisible by 7 starting from 50.
n = 50
while True:             # 'while True' runs forever unless we break
    if n % 7 == 0:
        print(f"First multiple of 7 from 50: {n}")
        break
    n += 1

print()

# --- while with continue ---
# Print numbers 1–10 but skip multiples of 3.
i = 0
print("1–10, skipping multiples of 3:")
while i < 10:
    i += 1
    if i % 3 == 0:
        continue        # skip the print, go back to the condition
    print(i, end=" ")
print()                 # newline after the inline prints

print()

# --- Simulating user input with a sentinel value ---
# In a real program you would use input(); here we simulate with a list.
simulated_inputs = ["hello", "world", "quit", "ignored"]
idx = 0
while True:
    user_input = simulated_inputs[idx]
    idx += 1
    if user_input == "quit":
        print("Exiting loop.")
        break
    print(f"You entered: {user_input}")
````

## 3. The `for` Loop and `range()`

A `for` loop **iterates over a sequence** (a list, string, range, etc.), executing its body once for each item.

```python
for variable in sequence:
    # body
```

### `range()` — generating integer sequences

| Call | Sequence produced | Common use |
|------|-------------------|------------|
| `range(5)` | `0, 1, 2, 3, 4` | Index from 0 |
| `range(1, 6)` | `1, 2, 3, 4, 5` | Count from 1 |
| `range(0, 10, 2)` | `0, 2, 4, 6, 8` | Even numbers |
| `range(10, 0, -1)` | `10, 9, …, 1` | Countdown |

**Remember:** `stop` is always **excluded**.

### Loop `else` clause

Both `for` and `while` support an optional `else` block that runs **only if the loop completed without hitting a `break`**. This is useful for "did we find it?" patterns.

````python
# --- range() variants ---
print("range(5):      ", list(range(5)))
print("range(1, 6):   ", list(range(1, 6)))
print("range(0,10,2): ", list(range(0, 10, 2)))
print("range(10,0,-1):", list(range(10, 0, -1)))

print()

# --- Iterating over a string ---
word = "Python"
print("Characters in 'Python':")
for char in word:
    print(char, end=" ")
print()

print()

# --- break inside a for loop ---
# Stop as soon as we find a negative number.
numbers = [4, 7, 2, -3, 8, -1]
print("Scanning for first negative number:")
for num in numbers:
    if num < 0:
        print(f"  Found: {num}")
        break
    print(f"  Checked {num} — positive, continuing...")

print()

# --- continue inside a for loop ---
# Print only odd numbers from 1 to 10.
print("Odd numbers 1–10:")
for n in range(1, 11):
    if n % 2 == 0:
        continue        # skip even numbers
    print(n, end=" ")
print()

print()

# --- Loop else: check whether a value was found ---
target = 99
data = [10, 25, 38, 47, 56]

for value in data:
    if value == target:
        print(f"Found {target}!")
        break
else:
    # This block runs only when the loop ended without break.
    print(f"{target} was not found in the list.")
````

## 4. Common Patterns

Control flow constructs are the backbone of several fundamental programming patterns you will use constantly.

### Accumulator pattern
Start with a base value (usually `0` for sums, `1` for products, `""` for strings) and update it on every iteration.

### Counter pattern
Start a counter at `0` and increment it each time a condition is met inside a loop.

### Finding min / max manually
Initialise a "best so far" variable with the first element, then compare every subsequent element against it, replacing it when you find a better candidate.

These patterns are the foundations of algorithms — understanding them deeply will help you write more complex logic later.

````python
data = [34, 7, 89, 12, 56, 3, 78, 45, 22, 61]

# --- Accumulator: sum all numbers ---
total = 0
for num in data:
    total += num        # accumulate: add each value to the running total
print(f"Sum of data:     {total}")

# --- Accumulator: build a string ---
letters = ["P", "y", "t", "h", "o", "n"]
word = ""
for letter in letters:
    word += letter
print(f"Built word:      {word}")

print()

# --- Counter: count numbers above a threshold ---
threshold = 40
above_count = 0
for num in data:
    if num > threshold:
        above_count += 1
print(f"Numbers > {threshold}:   {above_count}")

# --- Counter: count even numbers ---
even_count = 0
for num in data:
    if num % 2 == 0:
        even_count += 1
print(f"Even numbers:    {even_count}")

print()

# --- Finding minimum manually ---
minimum = data[0]       # assume first element is smallest
for num in data[1:]:    # check every other element
    if num < minimum:
        minimum = num   # found a new smallest — update
print(f"Minimum:         {minimum}")

# --- Finding maximum manually ---
maximum = data[0]
for num in data[1:]:
    if num > maximum:
        maximum = num
print(f"Maximum:         {maximum}")

print()

# --- Bonus: compute average using sum and counter together ---
running_sum = 0
running_count = 0
for num in data:
    running_sum += num
    running_count += 1
average = running_sum / running_count
print(f"Average:         {average:.1f}")
````
