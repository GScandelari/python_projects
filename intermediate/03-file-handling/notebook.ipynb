# File Handling in Python

Reading and writing files is one of the most common real-world tasks. Python's `with open(...)` pattern makes it safe and clean.

**In this notebook:**
- `open()` and the `with` statement
- Reading methods: `read`, `readlines`, iteration
- Writing and appending
- CSV and JSON files
- `pathlib.Path` for modern path handling

## 1. The `with` Statement

Always use `with open(...)` — it guarantees the file is closed even if an exception occurs.

```python
# Old way — risky
f = open("file.txt")
content = f.read()
f.close()   # skipped if an exception occurs above!

# Correct way
with open("file.txt", "r", encoding="utf-8") as f:
    content = f.read()
# file is closed here automatically
```

````python
# Create a sample file to work with
sample_text = """Python is amazing.
File handling is easy.
The with statement is your friend.
Always specify encoding.
Read before you write.
"""

with open("sample.txt", "w", encoding="utf-8") as f:
    f.write(sample_text)

print("File written successfully.")
````

## 2. Reading Files

Three approaches — choose based on file size and what you need:

````python
# read() — entire file as one string (good for small files)
with open("sample.txt", "r", encoding="utf-8") as f:
    content = f.read()
print(repr(content[:40]))   # show first 40 chars with escape chars

# readlines() — list of strings, newlines included
with open("sample.txt", "r", encoding="utf-8") as f:
    lines = f.readlines()
print(f"{len(lines)} lines")
print(lines[0])   # 'Python is amazing.\n'

# Iteration — most memory-efficient for large files
with open("sample.txt", "r", encoding="utf-8") as f:
    for i, line in enumerate(f, 1):
        print(f"{i:2}: {line.rstrip()}")
````

## 3. Writing and Appending

````python
# "w" — write (creates or OVERWRITES)
with open("output.txt", "w", encoding="utf-8") as f:
    f.write("Line 1\n")
    f.write("Line 2\n")

# "a" — append (adds to end, never overwrites)
with open("output.txt", "a", encoding="utf-8") as f:
    f.write("Line 3 — appended\n")

# writelines() — writes a list (you must include \n yourself)
extras = ["Line 4\n", "Line 5\n"]
with open("output.txt", "a", encoding="utf-8") as f:
    f.writelines(extras)

# Verify
with open("output.txt", "r", encoding="utf-8") as f:
    print(f.read())
````

## 4. CSV Files

Always use the `csv` module — never parse CSVs manually with `split(",")`. The module handles quoted fields, commas inside values, and different newline styles correctly.

````python
import csv

# Write CSV with DictWriter
students = [
    {"name": "Alice", "grade": 92, "city": "São Paulo"},
    {"name": "Bob",   "grade": 78, "city": "Rio de Janeiro"},
    {"name": "Carol", "grade": 85, "city": "Curitiba"},
]

with open("students.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "grade", "city"])
    writer.writeheader()
    writer.writerows(students)

# Read CSV with DictReader
with open("students.csv", "r", encoding="utf-8", newline="") as f:
    reader = csv.DictReader(f)
    for row in reader:
        status = "Pass" if int(row["grade"]) >= 70 else "Fail"
        print(f"{row['name']:<8} | {row['grade']:>3} | {status}")
````

## 5. JSON Files

````python
import json

# Write JSON
config = {
    "theme": "light",
    "font_size": 12,
    "language": "pt-BR",
    "recent_files": []
}

with open("config.json", "w", encoding="utf-8") as f:
    json.dump(config, f, indent=2, ensure_ascii=False)

# Read, modify, write back
with open("config.json", "r", encoding="utf-8") as f:
    loaded = json.load(f)

loaded["theme"] = "dark"
loaded["font_size"] = 14
loaded["recent_files"].append("notes.txt")

with open("config.json", "w", encoding="utf-8") as f:
    json.dump(loaded, f, indent=2, ensure_ascii=False)

# Print final state
for key, value in loaded.items():
    print(f"{key:<15}: {value}")
````

## 6. `pathlib.Path` — Modern Path Handling

`pathlib` replaces `os.path` with an object-oriented API. Paths are composable with `/`.

````python
from pathlib import Path

# Build paths with /
p = Path(".") / "sample.txt"
print(p.name)       # sample.txt
print(p.stem)       # sample
print(p.suffix)     # .txt
print(p.exists())   # True
print(p.resolve())  # absolute path

# Read and write directly
text = p.read_text(encoding="utf-8")
print(f"Characters: {len(text)}")

# List files by extension
current = Path(".")
py_files = list(current.glob("*.py"))     # current dir only
all_py = list(current.glob("**/*.py"))    # recursive
print(f"Local .py: {len(py_files)} | Total .py: {len(all_py)}")

# File info
for f in sorted(current.glob("*.txt")):
    size = f.stat().st_size
    print(f"{f.name:<20} {size:>6} bytes")
````
