# String Manipulation in Python

Strings are immutable sequences of characters. Python gives you a rich set of built-in methods to search, transform, format, and split them — with no imports needed.

**In this notebook:**
- Indexing and slicing
- Essential string methods
- f-string formatting
- Split, join, search
- Real-world patterns

## 1. Indexing and Slicing

Each character in a string has an index. Negative indices count from the end.

```
P  y  t  h  o  n
0  1  2  3  4  5
-6 -5 -4 -3 -2 -1
```

Slice syntax: `text[start:stop:step]` — stop is **exclusive**.

````python
text = "Python"

# Indexing
print(text[0])     # P
print(text[-1])    # n
print(text[2])     # t

# Slicing
print(text[1:4])   # yth
print(text[:3])    # Pyt  — from start
print(text[3:])    # hon  — to end
print(text[::2])   # Pto  — every 2nd character
print(text[::-1])  # nohtyP — reversed!

# Strings are immutable — you cannot assign to an index
# text[0] = 'p'  # TypeError!
new_text = 'p' + text[1:]  # create a new string instead
print(new_text)   # python
````

## 2. Case and Strip Methods

All string methods return a **new string** — the original is never modified.

````python
s = "  Hello, World!  "

# Case
print(s.upper())       # '  HELLO, WORLD!  '
print(s.lower())       # '  hello, world!  '
print(s.title())       # '  Hello, World!  '
print(s.swapcase())    # '  hELLO, wORLD!  '

# Strip
print(s.strip())       # 'Hello, World!'
print(s.lstrip())      # 'Hello, World!  '
print(s.rstrip())      # '  Hello, World!'

# Chaining — each method returns a string, so you can chain
clean = s.strip().lower().replace(",", "")
print(clean)           # 'hello world!'

# Check methods
print("123".isdigit())            # True
print("abc".isalpha())            # True
print("hello".startswith("he"))   # True
print("file.py".endswith(".py"))  # True
````

## 3. f-string Formatting

f-strings (Python 3.6+) are the most readable way to format strings. Use `{variable:format_spec}` inside the braces.

````python
name = "Alice"
age = 28
score = 95.6789
balance = 1_234_567.89

# Basic interpolation
print(f"Name: {name}, Age: {age}")

# Number formatting
print(f"Score: {score:.2f}")          # 2 decimal places
print(f"ID: {age:05d}")               # zero-padded integer
print(f"Balance: {balance:,.2f}")     # thousands separator

# Alignment
print(f"{'Name':<10} {'Score':>6}")   # left and right align
print(f"{'Alice':<10} {92:>6}")
print(f"{'Bob':<10} {78:>6}")

# Expression inside braces
print(f"2 + 2 = {2 + 2}")
print(f"Upper: {name.upper()}")
print(f"repr: {name!r}")              # includes quotes: 'Alice'
````

## 4. Searching and Replacing

````python
text = "the quick brown fox jumps over the lazy dog"

# Membership test — fastest way
print("fox" in text)              # True
print("cat" in text)              # False

# find() returns index, or -1 if not found
print(text.find("fox"))           # 16
print(text.find("cat"))           # -1

# count() — how many times a substring appears
print(text.count("the"))          # 2
print(text.count("o"))            # 4

# replace() — returns new string
updated = text.replace("fox", "cat")
print(updated)

# replace only first N occurrences
print(text.replace("the", "THE", 1))  # replaces only the first 'the'
````

## 5. Split and Join

**`split()`** breaks a string into a list.  
**`join()`** combines a list of strings into one.

They are inverses of each other and are used together constantly.

````python
# split()
sentence = "Python is easy to learn"
words = sentence.split()          # split on whitespace
print(words)   # ['Python', 'is', 'easy', 'to', 'learn']

csv = "Alice,28,Engineer"
fields = csv.split(",")
print(fields)  # ['Alice', '28', 'Engineer']

# split with maxsplit
print("a:b:c:d".split(":", 2))   # ['a', 'b', 'c:d']

# join()
words = ["Python", "is", "awesome"]
print(" ".join(words))    # Python is awesome
print("-".join(words))    # Python-is-awesome
print("".join(words))     # Pythonisawesome

# Classic pattern: split → transform → join
sentence = "hello world from python"
titled = " ".join(word.capitalize() for word in sentence.split())
print(titled)   # Hello World From Python

# join requires all items to be strings!
numbers = [1, 2, 3]
print(", ".join(str(n) for n in numbers))   # 1, 2, 3
````

## 6. Common Patterns

Real-world string tasks that combine everything above.

````python
# --- Palindrome check ---
def is_palindrome(text):
    clean = text.lower().replace(" ", "")
    return clean == clean[::-1]

print(is_palindrome("racecar"))   # True
print(is_palindrome("hello"))     # False

# --- Word frequency ---
text = "the cat sat on the mat the cat"
freq = {}
for word in text.split():
    freq[word] = freq.get(word, 0) + 1

for word, count in sorted(freq.items()):
    print(f"{word}: {count}")

# --- Input cleaning ---
raw_email = "  USER@EXAMPLE.COM  "
email = raw_email.strip().lower()
print(email)   # user@example.com

# --- Formatted table ---
data = [("Alice", 88), ("Bob", 72), ("Carol", 95)]
print(f"{'Name':<10} {'Score':>6}")
print("-" * 18)
for name, score in data:
    print(f"{name:<10} {score:>6}")
````
