# Data Structures in Python

Python provides four built-in collection types that cover the vast majority of everyday programming needs:

| Structure | Literal | Ordered | Mutable | Duplicates |
|-----------|---------|---------|---------|------------|
| **List** | `[1, 2, 3]` | Yes | Yes | Yes |
| **Tuple** | `(1, 2, 3)` | Yes | No | Yes |
| **Dictionary** | `{"key": val}` | Yes (3.7+) | Yes | Keys: No |
| **Set** | `{1, 2, 3}` | No | Yes | No |

By the end of this notebook you will know:
- How to create, read, and modify each structure
- The key methods that come with each type
- How to choose the right structure for a given problem

## 1. Lists

A **list** is an ordered, mutable sequence. It is the most versatile collection type and the one you will reach for most often.

### Creating lists

```python
empty  = []                        # empty list
fruits = ["apple", "banana", "cherry"]  # list of strings
mixed  = [1, "two", 3.0, True]    # lists can hold any types
nested = [[1, 2], [3, 4]]         # lists can contain other lists
```

### Indexing and slicing

```
 Index:  0       1        2
        ["apple", "banana", "cherry"]
Neg:    -3       -2        -1
```

- `lst[i]` — element at index `i` (0-based, negative from end)
- `lst[start:stop]` — slice from `start` up to (not including) `stop`
- `lst[start:stop:step]` — slice with a step

### Key list methods

| Method | What it does |
|--------|--------------|
| `.append(x)` | Add `x` to the end |
| `.insert(i, x)` | Insert `x` before index `i` |
| `.remove(x)` | Remove first occurrence of `x` |
| `.pop(i)` | Remove and return element at index `i` (default: last) |
| `.sort()` | Sort in place |
| `.reverse()` | Reverse in place |
| `.index(x)` | Return index of first occurrence of `x` |
| `.count(x)` | Count occurrences of `x` |
| `len(lst)` | Number of elements |

````python
# ---- Creating lists ----
fruits = ["apple", "banana", "cherry", "orange", "mango"]
numbers = [10, 30, 20, 50, 40]

print("fruits :", fruits)
print("numbers:", numbers)
print()

# ---- Indexing ----
print("=== Indexing ===")
print("First  (index  0):", fruits[0])    # apple
print("Third  (index  2):", fruits[2])    # cherry
print("Last   (index -1):", fruits[-1])   # mango
print("Second last (-2) :", fruits[-2])   # orange
print()

# ---- Slicing: lst[start:stop:step] ----
print("=== Slicing ===")
print("fruits[1:3]   :", fruits[1:3])     # ['banana', 'cherry']
print("fruits[:3]    :", fruits[:3])      # first 3  → ['apple', 'banana', 'cherry']
print("fruits[2:]    :", fruits[2:])      # from index 2 onward
print("fruits[::2]   :", fruits[::2])     # every other element
print("fruits[::-1]  :", fruits[::-1])    # reversed copy
print()

# ---- Key methods ----
print("=== Mutation methods ===")
basket = ["apple", "banana", "cherry"]
basket.append("mango")              # add to the end
print("after append('mango') :", basket)

basket.insert(1, "blueberry")       # insert before index 1
print("after insert(1, ...) :", basket)

basket.remove("banana")             # remove first 'banana'
print("after remove('banana'):", basket)

popped = basket.pop()               # remove and return last item
print("popped             :", popped)
print("basket after pop() :", basket)
print()

# ---- Sorting ----
print("=== Sorting ===")
nums = [5, 2, 8, 1, 9, 3]
nums.sort()                         # in-place ascending
print("sorted ascending :", nums)
nums.sort(reverse=True)             # in-place descending
print("sorted descending:", nums)
print("sorted() (new)   :", sorted([5, 2, 8, 1]))   # returns new list
print()

# ---- Other useful operations ----
print("=== Other ===")
colors = ["red", "green", "blue", "green"]
print("len     :", len(colors))          # 4
print("count   :", colors.count("green")) # 2
print("index   :", colors.index("blue")) # 2
print("in      :", "red" in colors)      # True
print("not in  :", "pink" not in colors) # True
````

## 2. Tuples

A **tuple** is an ordered, **immutable** sequence. Once created, its contents cannot change.

### When to use a tuple instead of a list

- The data is fixed by design (e.g., GPS coordinates, RGB colour values, database rows)
- You want to signal to readers that these values should not change
- You need to use the collection as a **dictionary key** (requires hashability → requires immutability)
- Returning multiple values from a function (`return x, y` creates a tuple)

### Creating tuples

```python
point     = (4, -2, 7)     # parentheses optional but conventional
single    = (42,)           # trailing comma is REQUIRED for one-element tuples
rgb       = 255, 128, 0    # parentheses can be omitted
```

> **Common mistake:** `(42)` is just the integer `42` in parentheses — it is NOT a tuple. You must write `(42,)`.

````python
# ---- Creating tuples ----
point = (4, -2, 7)
rgb   = (255, 128, 0)
empty = ()

print("point :", point)
print("rgb   :", rgb)
print("type  :", type(point))   # <class 'tuple'>
print()

# ---- One-element tuple — trailing comma is mandatory ----
not_a_tuple = (42)     # just the integer 42
is_a_tuple  = (42,)    # this IS a tuple
print("(42)  type:", type(not_a_tuple))   # <class 'int'>
print("(42,) type:", type(is_a_tuple))    # <class 'tuple'>
print()

# ---- Indexing works the same as lists ----
print("=== Indexing ===")
print("point[0] :", point[0])    # 4
print("point[-1]:", point[-1])   # 7
print("rgb[:2]  :", rgb[:2])     # (255, 128)
print()

# ---- Unpacking ----
print("=== Unpacking ===")
x, y, z = point                  # assign all three at once
print(f"x={x}, y={y}, z={z}")

red, green, blue = rgb
print(f"R={red}, G={green}, B={blue}")

# Swap two variables using tuple unpacking (a classic Python idiom)
a, b = 10, 20
print(f"before swap: a={a}, b={b}")
a, b = b, a
print(f"after swap : a={a}, b={b}")
print()

# ---- Immutability demo ----
print("=== Immutability ===")
coords = (10, 20)
print("coords       :", coords)
try:
    coords[0] = 99   # this will raise TypeError
except TypeError as e:
    print("Cannot modify tuple:", e)

# To 'change' a tuple you must create a new one
coords = (99, coords[1])
print("new coords   :", coords)
````

## 3. Dictionaries

A **dictionary** (`dict`) stores data as **key → value** pairs. Keys must be unique and hashable (strings, numbers, and tuples work; lists do not).

### Creating dictionaries

```python
person = {"name": "Alice", "age": 30}    # literal
empty  = {}                               # empty dict (NOT empty set!)
built  = dict(name="Alice", age=30)       # keyword arguments
```

### Essential methods

| Method | Returns | Notes |
|--------|---------|-------|
| `d[key]` | value | `KeyError` if missing |
| `d.get(key, default)` | value or default | Safe access, no exception |
| `d[key] = value` | — | Add or overwrite |
| `del d[key]` | — | Remove key |
| `d.keys()` | view of keys | Iterable, not a list |
| `d.values()` | view of values | Iterable |
| `d.items()` | view of (k, v) pairs | Best for iteration |
| `key in d` | bool | Membership test on keys |
| `d.pop(key)` | removed value | `KeyError` if missing |
| `d.update(other)` | — | Merge another dict in |

````python
# ---- Creating a dictionary ----
person = {
    "name" : "Alice",
    "age"  : 30,
    "city" : "São Paulo",
    "score": 95.5,
}

print("person:", person)
print("type  :", type(person))
print()

# ---- Accessing values ----
print("=== Access ===")
print("name :", person["name"])           # direct access — KeyError if missing
print("age  :", person["age"])
print("phone:", person.get("phone"))      # safe — returns None
print("phone:", person.get("phone", "N/A"))  # with default value
print()

# ---- Modifying a dictionary ----
print("=== Modify ===")
person["age"]   = 31                      # update existing key
person["email"] = "alice@example.com"     # add new key
print("after update:", person)

removed = person.pop("score")             # remove and return value
print("removed score:", removed)
print("after pop    :", person)
print()

# ---- Views: keys / values / items ----
print("=== Views ===")
print("keys  :", list(person.keys()))
print("values:", list(person.values()))
print("items :", list(person.items()))
print()

# ---- Iteration ----
print("=== Iteration ===")

# Iterate over keys (default behaviour)
for key in person:
    print(f"  {key}")

print()

# Iterate over key-value pairs with .items() — the most useful pattern
for key, value in person.items():
    print(f"  {key:8s}: {value}")

print()

# ---- Membership test ----
print("=== Membership ===")
print('"name" in person  :', "name" in person)    # True
print('"phone" in person :', "phone" in person)   # False

# ---- Merging two dicts (Python 3.9+ supports | operator) ----
defaults = {"theme": "light", "lang": "en"}
overrides = {"lang": "pt-BR", "font": "mono"}
merged = {**defaults, **overrides}   # spread operator — works in all 3.x
print()
print("merged:", merged)
````

## 4. Sets

A **set** is an unordered collection of **unique** elements. It is backed by a hash table, which makes membership tests (`in`) extremely fast — O(1) on average.

### Creating sets

```python
colors   = {"red", "green", "blue"}   # literal
from_lst = set([1, 2, 2, 3, 3, 3])   # deduplicate a list → {1, 2, 3}
empty    = set()                      # NOT {} — that creates an empty dict!
```

### Key operations

| Operation | Method | Operator |
|-----------|--------|----------|
| Add element | `.add(x)` | — |
| Remove (error if missing) | `.remove(x)` | — |
| Remove (silent if missing) | `.discard(x)` | — |
| Union | `.union(other)` | `a \| b` |
| Intersection | `.intersection(other)` | `a & b` |
| Difference | `.difference(other)` | `a - b` |
| Symmetric difference | `.symmetric_difference(other)` | `a ^ b` |
| Subset check | `.issubset(other)` | `a <= b` |
| Superset check | `.issuperset(other)` | `a >= b` |

````python
# ---- Creating sets ----
colors = {"red", "green", "blue"}
print("colors:", colors)
print("type  :", type(colors))
print()

# ---- Deduplication: build a set from a list with duplicates ----
raw = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5]
unique = set(raw)
print("raw   :", raw, "  length:", len(raw))
print("unique:", unique, "length:", len(unique))
print()

# ---- Add and remove elements ----
print("=== Add / Remove ===")
fruits = {"apple", "banana", "cherry"}
fruits.add("mango")          # add one element
fruits.add("apple")          # already there — silently ignored
print("after add     :", fruits)

fruits.discard("banana")     # remove if present (no error if missing)
fruits.discard("durian")     # not present — no error
print("after discard :", fruits)

try:
    fruits.remove("durian")  # remove — raises KeyError if missing
except KeyError as e:
    print(f"remove() raised KeyError: {e}")

print()

# ---- Membership test — O(1) ----
print("=== Membership ===")
print('"apple" in fruits  :', "apple" in fruits)    # True
print('"banana" in fruits :', "banana" in fruits)   # False
print()

# ---- Set operations ----
print("=== Set operations ===")
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

print("a          :", a)
print("b          :", b)
print()
print("union      :", a | b)   # all elements from both
print("intersection:", a & b)  # only elements in both
print("difference :", a - b)   # in a but not in b
print("sym. diff  :", a ^ b)   # in one but not both
print()

# ---- Subset / superset ----
print("=== Subset / Superset ===")
small = {1, 2}
big   = {1, 2, 3, 4, 5}
print("small <= big :", small <= big)    # True — small is subset
print("big >= small :", big >= small)    # True — big is superset
print("small == big :", small == big)    # False
````

## 5. Choosing the Right Structure

Picking the correct data structure is one of the most important skills in programming. Here is a practical guide:

| Question | Answer → Structure |
|----------|--------------------|
| Do I need a simple ordered sequence that can change? | **List** |
| Do I have a fixed record (coordinates, config, return values)? | **Tuple** |
| Do I need to look things up by a name/label/key? | **Dictionary** |
| Do I only care about membership, or need unique values? | **Set** |

### Full comparison table

| Feature | `list` | `tuple` | `dict` | `set` |
|---------|--------|---------|--------|-------|
| Ordered | Yes | Yes | Yes (3.7+) | No |
| Mutable | Yes | No | Yes | Yes |
| Duplicates | Yes | Yes | Keys: No / Values: Yes | No |
| Key-value pairs | No | No | Yes | No |
| Index access | Yes | Yes | By key | No |
| Hashable (dict key) | No | Yes (if contents hashable) | No | No |
| Membership `in` | O(n) | O(n) | O(1) | O(1) |
| Typical use | Ordered sequence | Fixed record | Named data | Unique items |

````python
# ---- Side-by-side examples showing when to use each structure ----

# SCENARIO 1: Shopping cart — ordered list of items that changes over time
# → Use a LIST (ordered, mutable, duplicates allowed)
cart = ["apple", "bread", "milk", "apple"]   # apple appears twice — that's fine
cart.append("eggs")
cart.remove("bread")
print("Shopping cart (list):", cart)
print()

# SCENARIO 2: GPS waypoint — fixed (latitude, longitude, altitude)
# → Use a TUPLE (fixed record, immutable, can be a dict key)
home_coords = (-23.5505, -46.6333, 760)   # São Paulo
office_coords = (-23.5613, -46.6558, 850)
waypoints = {home_coords: "Home", office_coords: "Office"}  # tuple as dict key
print("Waypoint lookup (tuple as key):", waypoints[home_coords])
print()

# SCENARIO 3: User profile — named fields that you look up by key
# → Use a DICTIONARY (key-value pairs, fast lookup)
user = {
    "username": "alice99",
    "email"   : "alice@example.com",
    "age"     : 30,
    "premium" : True,
}
print(f"User profile (dict): {user['username']} — premium: {user['premium']}")
print()

# SCENARIO 4: Visited pages — track which URLs a user has already seen
# → Use a SET (membership test only, no duplicates, O(1) lookup)
visited = {"/home", "/about", "/products"}
current_page = "/about"
if current_page in visited:
    print(f"Visited pages (set): {current_page!r} already visited — skip fetch")
visited.add("/contact")
print("Visited set:", visited)
print()

# SCENARIO 5: Deduplicate a list while preserving order (common pattern)
# → Convert to set to remove duplicates, then sort or use dict.fromkeys
raw_tags = ["python", "data", "python", "tutorial", "data", "beginner"]
unique_tags = list(dict.fromkeys(raw_tags))   # preserves first-seen order
print("Raw tags   :", raw_tags)
print("Unique tags:", unique_tags)            # order preserved
print("Via set    :", list(set(raw_tags)))    # order NOT preserved
````
