## 1. Variables

A **variable** is a named container for a value. In Python you create one with a single `=` sign:

```python
name = value
```

Python is **dynamically typed** — you do not need to declare the type in advance. The interpreter figures it out from the value you assign.

### The five primitive types

| Type | Keyword | Example literal |
|------|---------|----------------|
| Integer | `int` | `42` |
| Floating-point | `float` | `3.14` |
| String | `str` | `"hello"` |
| Boolean | `bool` | `True` / `False` |
| Nothing | `NoneType` | `None` |

> **Naming rules:** variable names must start with a letter or underscore, can contain letters, digits and underscores, and are case-sensitive (`age` and `Age` are different variables). Use `snake_case` by convention.

````python
# --- Integer: whole numbers (positive, negative, or zero) ---
age = 25
year = 2026
temperature_below_zero = -8

# --- Float: numbers with a decimal point ---
height = 1.75
pi = 3.14159
small = 0.001

# --- String: text enclosed in quotes (single or double, both work) ---
name = "Alice"
greeting = 'Hello, World!'

# --- Boolean: only two possible values, always capitalised ---
is_student = True
has_passed  = False

# --- NoneType: represents the absence of a value ---
result = None

# Print every variable so we can see its value
print("age            :", age)
print("year           :", year)
print("height         :", height)
print("name           :", name)
print("greeting       :", greeting)
print("is_student     :", is_student)
print("result         :", result)

# type() tells you exactly what Python thinks a variable is
print()
print("type(age)      :", type(age))
print("type(height)   :", type(height))
print("type(name)     :", type(name))
print("type(is_student):", type(is_student))
print("type(result)   :", type(result))
````

## 2. Operators

Operators let you compute new values from existing ones.

### Arithmetic operators

| Operator | Symbol | Example | Result |
|----------|--------|---------|--------|
| Addition | `+` | `3 + 2` | `5` |
| Subtraction | `-` | `3 - 2` | `1` |
| Multiplication | `*` | `3 * 2` | `6` |
| Division | `/` | `7 / 2` | `3.5` (always float) |
| Floor division | `//` | `7 // 2` | `3` (drops remainder) |
| Modulo | `%` | `7 % 2` | `1` (remainder only) |
| Exponentiation | `**` | `3 ** 2` | `9` |

### Comparison operators

These always return a `bool` (`True` or `False`).

| Operator | Meaning |
|----------|---------|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |

````python
a = 18
b = 5

# ---- Arithmetic ----
print("=== Arithmetic ===")
print(f"{a} + {b}  = {a + b}")    # Addition
print(f"{a} - {b}  = {a - b}")    # Subtraction
print(f"{a} * {b}  = {a * b}")    # Multiplication
print(f"{a} / {b}  = {a / b}")    # Division — note the float result
print(f"{a} // {b} = {a // b}")   # Floor division — whole part only
print(f"{a} % {b}  = {a % b}")    # Modulo — remainder
print(f"{a} ** {b} = {a ** b}")   # Exponentiation

print()

# ---- Comparison — always produce True or False ----
print("=== Comparison ===")
print(f"{a} == {b}  →", a == b)   # Equal?
print(f"{a} != {b}  →", a != b)   # Not equal?
print(f"{a} >  {b}  →", a > b)    # Greater than?
print(f"{a} <  {b}  →", a < b)    # Less than?
print(f"{a} >= {b}  →", a >= b)   # Greater than or equal?
print(f"{a} <= {b}  →", a <= b)   # Less than or equal?

print()

# ---- Augmented assignment operators ----
# These update a variable in place and are very common in loops.
score = 10
score += 5     # same as: score = score + 5
print("score after += 5 :", score)
score *= 2     # same as: score = score * 2
print("score after *= 2 :", score)
score -= 7     # same as: score = score - 7
print("score after -= 7 :", score)
score /= 3     # same as: score = score / 3 — becomes float!
print("score after /= 3 :", score)
````

## 3. Type Conversion

Sometimes you need to convert a value from one type to another. Python provides built-in constructor functions for this:

| Function | Converts to | Notes |
|----------|------------|-------|
| `int(x)` | Integer | Truncates floats; strings must look like whole numbers |
| `float(x)` | Float | Integers become e.g. `3 → 3.0` |
| `str(x)` | String | Works on almost anything |
| `bool(x)` | Boolean | `0`, `0.0`, `""`, `None`, empty collections → `False`; everything else → `True` |

**The most common beginner error** is trying to concatenate (`+`) a number with a string without converting first:

```python
# This raises TypeError:
print("I am " + 25 + " years old")    # WRONG

# Fix with str():
print("I am " + str(25) + " years old")  # OK (but f-strings are nicer)
```

````python
# ---- int() conversions ----
print("=== int() ===")
print(int(3.9))        # 3   — truncates (does NOT round)
print(int(3.1))        # 3   — truncates toward zero
print(int("42"))       # 42  — string that looks like a whole number
print(int(True))       # 1   — True is 1 in Python
print(int(False))      # 0   — False is 0 in Python

print()

# ---- float() conversions ----
print("=== float() ===")
print(float(7))        # 7.0
print(float("3.14"))   # 3.14 — string that looks like a float
print(float(True))     # 1.0

print()

# ---- str() conversions ----
print("=== str() ===")
print(str(100))        # '100'
print(str(3.14))       # '3.14'
print(str(True))       # 'True'
print(str(None))       # 'None'

print()

# ---- bool() — the falsy values ----
print("=== bool() — falsy values ===")
print(bool(0))         # False — zero int
print(bool(0.0))       # False — zero float
print(bool(""))        # False — empty string
print(bool(None))      # False — None
print()
print("=== bool() — truthy values ===")
print(bool(1))         # True
print(bool(-99))       # True — any non-zero number
print(bool("hello"))   # True — any non-empty string

print()

# ---- Verify conversions with type() ----
x = "123"
print("Before conversion:", x, "→", type(x))
x = int(x)
print("After  int(x)    :", x, "→", type(x))
x = float(x)
print("After  float(x)  :", x, "→", type(x))
x = str(x)
print("After  str(x)    :", x, "→", type(x))
````

## 4. Formatting Output

### `print()`

`print()` writes values to the screen. Key parameters:

- **`sep`** — separator between multiple values (default: a single space `' '`)
- **`end`** — what to print at the end (default: newline `'\n'`)

### f-strings (formatted string literals)

Introduced in Python 3.6, f-strings are the preferred way to embed values inside text. Prefix the string with `f` and place any expression inside `{}`:

```python
name = "Alice"
age  = 25
print(f"My name is {name} and I am {age} years old.")
# → My name is Alice and I am 25 years old.
```

You can also control formatting with a **format spec** inside the braces:

```python
pi = 3.14159
print(f"Pi to 2 decimal places: {pi:.2f}")
# → Pi to 2 decimal places: 3.14
```

### `input()`

`input(prompt)` pauses the program, displays the prompt, and returns whatever the user types as a **string**. Always convert when you need a number.

````python
# ---- Basic print() usage ----
print("Hello, World!")

# Print multiple values — separated by a space by default
print("Name:", "Alice", "Age:", 25)

# Custom separator
print("2026", "04", "11", sep="-")   # → 2026-04-11

# Custom end character (suppress the newline)
print("Loading", end="")
print("...", end="\n")              # → Loading...

print()

# ---- f-strings: embed variables directly in text ----
name = "Alice"
age  = 25
height = 1.75

# Basic variable interpolation
print(f"Name  : {name}")
print(f"Age   : {age}")

# Expressions work inside {} too
print(f"In 10 years, {name} will be {age + 10} years old.")

print()

# ---- Format specifiers for numbers ----
pi = 3.14159265
price = 1234.5

print(f"Pi (2 dp)       : {pi:.2f}")        # fixed 2 decimal places
print(f"Pi (4 dp)       : {pi:.4f}")        # fixed 4 decimal places
print(f"Price (comma)   : {price:,.2f}")    # thousands separator + 2 dp
print(f"Height          : {height:.0f} m") # round to 0 decimal places

print()

# ---- input() — reading from the user ----
# NOTE: The cell below is commented out so the notebook runs without pausing.
# Uncomment and run it interactively to see input() in action.

# user_name = input("What is your name? ")
# user_age  = int(input("How old are you? "))   # convert string → int!
# print(f"Hello, {user_name}! You are {user_age} years old.")
# print(f"In 5 years you will be {user_age + 5}.")

# ---- Demo without real input ----
# Simulating what input() would produce:
simulated_input = "42"                          # input() always returns str
number = int(simulated_input)                   # convert before doing maths
print(f"You entered: {simulated_input!r}  (type: {type(simulated_input).__name__})")
print(f"After int() : {number}            (type: {type(number).__name__})")
print(f"Double that : {number * 2}")
````
