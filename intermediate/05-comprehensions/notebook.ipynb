# List, Dict & Set Comprehensions

Comprehensions are a concise, Pythonic way to build collections from iterables. They replace many `for` loops in a single readable expression.

**In this notebook:**
- List comprehensions (transform + filter)
- if/else inside comprehensions
- Nested comprehensions
- Dict comprehensions
- Set comprehensions
- Generator expressions
- When NOT to use them

## 1. List Comprehension — basics

**Syntax:** `[expression for item in iterable]`

Replace a loop that builds a list by appending.

````python
# Loop version
squares_loop = []
for n in range(1, 6):
    squares_loop.append(n ** 2)

# Comprehension — same result
squares = [n ** 2 for n in range(1, 6)]

print(squares_loop)   # [1, 4, 9, 16, 25]
print(squares)        # [1, 4, 9, 16, 25]
print(squares_loop == squares)   # True

# Works on any iterable
names = ['alice', 'bob', 'carol']
upper = [name.upper() for name in names]
lengths = [len(name) for name in names]
print(upper)     # ['ALICE', 'BOB', 'CAROL']
print(lengths)   # [5, 3, 5]
````

## 2. Filtering with `if`

**Syntax:** `[expression for item in iterable if condition]`

The `if` at the end **filters** — elements that don't match are dropped.

````python
numbers = list(range(1, 11))

evens = [n for n in numbers if n % 2 == 0]
print(evens)   # [2, 4, 6, 8, 10]

# Combine: transform AND filter
even_squares = [n ** 2 for n in numbers if n % 2 == 0]
print(even_squares)   # [4, 16, 36, 64, 100]

# Filter strings by length
words = ['hi', 'hello', 'hey', 'world', 'python']
long_words = [w for w in words if len(w) > 3]
print(long_words)   # ['hello', 'world', 'python']
````

## 3. if/else inside a comprehension

When `if/else` controls the **value** (not which items appear), it goes *before* the `for`:

```
[value_if_true if condition else value_if_false  for item in iterable]
```

````python
# Every element stays — just its value changes
labels = ['even' if n % 2 == 0 else 'odd' for n in range(1, 7)]
print(labels)   # ['odd', 'even', 'odd', 'even', 'odd', 'even']

# Combine ternary (value) + filter (which items appear)
result = [n * 2 if n % 2 == 0 else n for n in range(1, 11) if n > 4]
print(result)   # [5, 12, 7, 16, 9, 20]

# Clamp values to a range
raw = [-3, 5, 12, 0, 7, -1, 10]
clamped = [max(0, min(10, v)) for v in raw]
print(clamped)   # [0, 5, 10, 0, 7, 0, 10]
````

## 4. Nested Comprehensions

A second `for` clause iterates over each sub-iterable.

````python
# Flatten a 2D matrix
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat = [val for row in matrix for val in row]
print(flat)   # [1, 2, 3, 4, 5, 6, 7, 8, 9]

# Cartesian product
pairs = [(x, y) for x in [1, 2, 3] for y in ['a', 'b']]
print(pairs)
# [(1,'a'),(1,'b'),(2,'a'),(2,'b'),(3,'a'),(3,'b')]

# Build a 2D table (list of lists)
table = [[row * col for col in range(1, 4)] for row in range(1, 4)]
for row in table:
    print(row)
# [1, 2, 3]
# [2, 4, 6]
# [3, 6, 9]
````

## 5. Dict Comprehensions

**Syntax:** `{key_expr: value_expr for item in iterable}`

````python
# Square lookup table
squares = {n: n ** 2 for n in range(1, 6)}
print(squares)   # {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}

# Invert a dict (swap keys and values)
original = {'a': 1, 'b': 2, 'c': 3}
inverted = {v: k for k, v in original.items()}
print(inverted)   # {1: 'a', 2: 'b', 3: 'c'}

# Filter while building
words = ['hello', 'world', 'python', 'hi']
long_words = {w: len(w) for w in words if len(w) > 3}
print(long_words)   # {'hello': 5, 'world': 5, 'python': 6}

# Normalize keys
keys = ['Name', 'AGE', 'CITY']
normalized = {k.lower(): None for k in keys}
print(normalized)   # {'name': None, 'age': None, 'city': None}
````

## 6. Set Comprehensions

**Syntax:** `{expression for item in iterable}`

Duplicates are automatically removed — order is not guaranteed.

````python
# Unique squared values
numbers = [1, -1, 2, -2, 3]
unique_squares = {n ** 2 for n in numbers}
print(unique_squares)   # {1, 4, 9}

# Unique first characters
words = ['apple', 'ant', 'banana', 'avocado', 'cherry']
first_chars = {w[0] for w in words}
print(first_chars)   # {'a', 'b', 'c'}  (order varies)

# Deduplicate while filtering
raw = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3]
unique_evens = {n for n in raw if n % 2 == 0}
print(unique_evens)   # {2, 4, 6}
````

## 7. Generator Expressions

Replace `[...]` with `(...)` to get a **lazy generator** — values computed on demand, no list built in memory.

````python
import sys

# Memory comparison
list_version = [n ** 2 for n in range(10_000)]
gen_version  = (n ** 2 for n in range(10_000))

print(f'List size : {sys.getsizeof(list_version):,} bytes')
print(f'Gen size  : {sys.getsizeof(gen_version):,} bytes')

# sum() works with generators directly
total = sum(n ** 2 for n in range(1_000_000))
print(f'Sum of squares: {total:,}')

# next() to get the first match lazily
first = next(n for n in range(1, 10_000) if n % 7 == 0 and n % 11 == 0)
print(f'First multiple of 7 and 11: {first}')   # 77

# all() and any() short-circuit with generators
nums = [2, 4, 6, 8, 10]
print(all(n % 2 == 0 for n in nums))   # True
print(any(n > 8 for n in nums))        # True
````

## 8. When NOT to use comprehensions

Comprehensions shine for **simple** transformations. When logic grows complex, a regular `for` loop is clearer.

````python
data = [1, 2, 3, 4, 5]

# AVOID: side effects inside a comprehension
# [print(x) for x in data]  # works but is bad style — use a loop

# AVOID: deeply nested (hard to read)
# nested = [[cell*2 for cell in row if cell > 2] for row in matrix if len(row) > 2]

# PREFER: extract a helper function for complex per-element logic
def process(x):
    doubled = x * 2
    if doubled > 5:
        return doubled
    return None

result = [process(x) for x in data if process(x) is not None]
print(result)   # [6, 8, 10]

# Better still — compute once, filter after
processed = [process(x) for x in data]
result = [v for v in processed if v is not None]
print(result)   # [6, 8, 10]
````
