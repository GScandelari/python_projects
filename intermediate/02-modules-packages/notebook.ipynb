# Modules and Packages

A **module** is any `.py` file. A **package** is a folder with an `__init__.py`. They let you split code across files, reuse logic, and tap into Python's 200+ standard library modules.

**In this notebook:**
- Import styles
- `math`, `random`, `datetime`, `os`, `json`
- Creating your own module
- Package structure and `__init__.py`
- The `__name__` guard

## 1. Import Styles

| Style | Usage | Notes |
|---|---|---|
| `import math` | `math.sqrt(9)` | Explicit, no collisions |
| `import math as m` | `m.sqrt(9)` | Short alias |
| `from math import sqrt` | `sqrt(9)` | Concise, may clash |
| `from math import *` | `sqrt(9)` | Avoid — pollutes namespace |

````python
import math
import math as m
from math import sqrt, pi

print(math.sqrt(144))    # 12.0
print(m.pi)              # 3.141592653589793
print(sqrt(25))          # 5.0
print(pi)                # 3.141592653589793

# Explore what's inside a module
public = [name for name in dir(math) if not name.startswith('_')]
print(public[:10])       # first 10 public names
````

## 2. `math` and `random`

````python
import math
import random

# math
print(math.ceil(4.2))       # 5
print(math.floor(4.8))      # 4
print(math.log(100, 10))    # 2.0
print(math.factorial(6))    # 720

# random
random.seed(42)             # reproducible
print(random.randint(1, 10))              # random int 1-10
print(random.choice(["rock", "paper", "scissors"]))
print(sorted(random.sample(range(1, 51), 5)))  # 5 unique ints

deck = list(range(1, 14))
random.shuffle(deck)
print(deck)
````

## 3. `datetime`

````python
from datetime import date, datetime, timedelta

today = date.today()
now = datetime.now()

print(today)                                  # 2026-04-11
print(today.strftime("%d/%m/%Y"))             # 11/04/2026
print(now.strftime("%A, %d %B %Y %H:%M"))    # Saturday, 11 April 2026 14:30

# Date arithmetic
in_30_days = today + timedelta(days=30)
print(in_30_days)

# Parse a string into a date
parsed = datetime.strptime("25/12/2025", "%d/%m/%Y")
print(parsed.date())

# Days between two dates
new_year = date(2027, 1, 1)
print(f"Days until 2027: {(new_year - today).days}")
````

## 4. `os` and `json`

````python
import os
import json

# os
print(os.getcwd())
print(os.path.join("folder", "sub", "file.txt"))  # OS-aware path
print(os.path.exists("README.md"))

# json — dict ↔ string
data = {"name": "Alice", "scores": [90, 85, 92], "active": True}

json_str = json.dumps(data, indent=2)
print(json_str)

parsed = json.loads(json_str)
print(type(parsed))           # <class 'dict'>
print(parsed["scores"])       # [90, 85, 92]
````

## 5. Creating Your Own Module

Any `.py` file is a module. The `__name__` guard prevents code from running on import.

```python
# mathutils.py
PI = 3.14159

def circle_area(radius):
    return PI * radius ** 2

def is_prime(n):
    if n < 2: return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0: return False
    return True

if __name__ == "__main__":
    # Only runs when: python mathutils.py
    # Skipped when:   import mathutils
    print(circle_area(5))
    print(is_prime(17))
```

````python
# Demonstrate __name__
print(f"This notebook's __name__: {__name__}")
# In a script run directly: __main__
# In a module being imported: the module's filename

# The guard pattern
def main():
    print("Running as a script!")

if __name__ == "__main__":
    main()   # runs if this file is executed directly
````

## 6. Package Structure

A package is a folder with `__init__.py`:

```
myutils/
├── __init__.py      ← exposes public API
├── strings.py       ← string utilities
└── numbers.py       ← numeric utilities
```

```python
# __init__.py — clean public API
from .strings import reverse, is_palindrome
from .numbers import clamp, is_prime

__all__ = ["reverse", "is_palindrome", "clamp", "is_prime"]
```

Users import cleanly:
```python
from myutils import reverse, is_prime
```

The dot in `from .strings import` is a **relative import** — it imports from within the same package.
