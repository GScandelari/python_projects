## 1. Series and DataFrame

```python
import pandas as pd
import numpy as np

# Series — 1-D labelled array
s = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
print(s['b'])    # 20

# DataFrame
df = pd.DataFrame({
    'name':  ['Alice', 'Bob', 'Carol', 'David'],
    'age':   [25, 30, 22, 28],
    'score': [88.5, 74.0, 92.3, 67.1],
    'dept':  ['Eng', 'Sales', 'Eng', 'Sales'],
})
print(df)
print(df.shape)    # (4, 4)
print(df.dtypes)
```

## 2. loc / iloc Indexing

```python
# loc — label-based
print(df.loc[0])                       # first row by label
print(df.loc[0:2, 'name'])             # rows 0–2, column 'name'
print(df.loc[df['age'] > 25])          # boolean filter

# iloc — position-based
print(df.iloc[0])          # first row
print(df.iloc[-1])         # last row
print(df.iloc[0:2, 1:3])   # rows 0–1, cols 1–2
```

## 3. Filtering and Querying

```python
# Boolean conditions — use & and | (not and/or)
print(df[df['score'] >= 80])
print(df[(df['age'] > 24) & (df['dept'] == 'Eng')])

# query() — cleaner for complex filters
print(df.query('age > 24 and dept == "Eng"'))

# String methods
print(df[df['name'].str.startswith('A')])
print(df[df['name'].str.contains('l', case=False)])
```

## 4. Adding and Transforming Columns

```python
# Vectorised operation — preferred over apply for math
df['score_norm'] = (df['score'] - df['score'].min()) / \
                   (df['score'].max() - df['score'].min())

# apply — row-wise custom function
df['grade'] = df['score'].apply(lambda x: 'A' if x >= 90 else ('B' if x >= 80 else 'C'))

print(df[['name', 'score', 'score_norm', 'grade']])
```

## 5. groupby

```python
# Aggregate by group
print(df.groupby('dept')['score'].mean())
print(df.groupby('dept').agg({'score': 'mean', 'age': 'max'}))

# transform — returns same-size result aligned to original index
df['dept_avg'] = df.groupby('dept')['score'].transform('mean')
print(df[['name', 'dept', 'score', 'dept_avg']])
```

## 6. Missing Data

```python
df_miss = pd.DataFrame({
    'name':  ['Ana', 'Bruno', 'Clara', 'Diego'],
    'age':   [28, None, 35, 29],
    'score': [90, 75, None, 88],
})

print(df_miss.isnull().sum())

# Drop rows where score is missing
clean = df_miss.dropna(subset=['score'])

# Fill numeric NaN with median
df_miss['age'] = df_miss['age'].fillna(df_miss['age'].median())
print(df_miss)
```

## 7. Merging DataFrames

```python
customers = pd.DataFrame({
    'id':   [1, 2, 3, 4],
    'name': ['Alice', 'Bob', 'Carol', 'David'],
})
orders = pd.DataFrame({
    'id':    [2, 3, 3],
    'total': [150, 320, 80],
})

inner = pd.merge(customers, orders, on='id', how='inner')
print('Inner:')
print(inner)

left = pd.merge(customers, orders, on='id', how='left')
left['total'] = left['total'].fillna(0)
print('\nLeft (with 0 fill):')
print(left)
```

## 8. Time Series Basics

```python
np.random.seed(42)
bdays = pd.bdate_range('2024-01-01', periods=20)
returns = np.random.normal(0, 0.01, 20)
price = pd.Series(100 * (1 + returns).cumprod(), index=bdays)

print(price.head())
print('Max:', price.idxmax().date(), price.max().round(2))

# Rolling mean
print(price.rolling(5).mean().round(2))

# Resample to weekly
print(price.resample('W').last().round(2))
```
