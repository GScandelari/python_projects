## 1. Figure and Axes

Always use `fig, ax = plt.subplots()`. The `ax` object gives full control over every element.

````python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(7, 4))
ax.set_title('Empty canvas')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

## 2. Line Plot

````python
x = np.linspace(0, 2 * np.pi, 200)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x, np.sin(x), color='steelblue', linewidth=2, label='sin(x)')
ax.plot(x, np.cos(x), color='tomato', linestyle='--', linewidth=2, label='cos(x)')
ax.axhline(0, color='gray', linewidth=0.8, linestyle=':')
ax.set_title('Trig Functions')
ax.set_xlabel('x (radians)')
ax.set_ylabel('Amplitude')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

## 3. Scatter Plot

````python
np.random.seed(0)
x = np.random.uniform(0, 10, 100)
y = 1.5 * x + np.random.randn(100) * 2

fig, ax = plt.subplots(figsize=(7, 5))
sc = ax.scatter(x, y, c=y, cmap='viridis', alpha=0.7,
                edgecolors='k', linewidths=0.4, s=60)
plt.colorbar(sc, ax=ax, label='y value')
ax.set_title('Linear Relationship')
ax.set_xlabel('x')
ax.set_ylabel('y')
plt.tight_layout()
plt.show()
````

## 4. Bar Chart

````python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales  = [320, 285, 410, 390, 450, 370]

fig, ax = plt.subplots(figsize=(8, 4))
bars = ax.bar(months, sales, color='steelblue', edgecolor='k', linewidth=0.5)
for bar, val in zip(bars, sales):
    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 5,
            str(val), ha='center', va='bottom', fontsize=9)
ax.set_title('Monthly Sales')
ax.set_ylabel('Units Sold')
ax.set_ylim(0, max(sales) * 1.15)
plt.tight_layout()
plt.show()
````

## 5. Histogram

````python
np.random.seed(42)
heights = np.random.normal(170, 10, 1000)

fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(heights, bins=30, color='steelblue', edgecolor='white', alpha=0.85)
ax.axvline(heights.mean(), color='red', linestyle='--', linewidth=1.5,
           label=f'Mean = {heights.mean():.1f}')
ax.set_title('Height Distribution')
ax.set_xlabel('Height (cm)')
ax.set_ylabel('Count')
ax.legend()
plt.tight_layout()
plt.show()
````

## 6. Subplots

````python
x = np.linspace(0, 2 * np.pi, 200)
np.random.seed(1)

fig, axes = plt.subplots(2, 2, figsize=(10, 7))

axes[0, 0].plot(x, np.sin(x), color='steelblue')
axes[0, 0].set_title('sin(x)')

axes[0, 1].plot(x, np.cos(x), color='tomato', linestyle='--')
axes[0, 1].set_title('cos(x)')

axes[1, 0].scatter(np.random.randn(60), np.random.randn(60), alpha=0.6)
axes[1, 0].set_title('Scatter')

axes[1, 1].hist(np.random.exponential(2, 300), bins=20, color='mediumseagreen')
axes[1, 1].set_title('Exponential dist.')

fig.suptitle('Overview', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
````

## 7. Annotations and Customisation

````python
np.random.seed(2)
t   = np.linspace(0, 4 * np.pi, 200)
sig = 2 * np.sin(t) + np.random.randn(200) * 0.3
peak_idx = np.argmax(sig)

fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(t, sig, color='steelblue', linewidth=1.2)
ax.axhline(sig.mean(), color='gray', linestyle='--', linewidth=1,
           label=f'Mean = {sig.mean():.2f}')
ax.fill_between(t, sig.mean() - sig.std(), sig.mean() + sig.std(),
                alpha=0.15, color='orange', label='Mean ± 1 std')
ax.annotate('Peak',
            xy=(t[peak_idx], sig[peak_idx]),
            xytext=(t[peak_idx] + 1, sig[peak_idx] - 0.7),
            arrowprops=dict(arrowstyle='->', color='red'),
            fontsize=9, color='red')
ax.set_title('Annotated Signal')
ax.legend(fontsize=8)
ax.grid(True, alpha=0.2)
plt.tight_layout()
plt.show()
````

## 8. Saving Figures

```python
fig.savefig('plot.png', dpi=150, bbox_inches='tight')
fig.savefig('plot.svg')   # vector — scales without blur
fig.savefig('plot.pdf')   # publication quality
plt.close(fig)            # free memory
```
