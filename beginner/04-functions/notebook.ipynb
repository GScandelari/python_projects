# Functions in Python

Functions are reusable blocks of code that perform a specific task. They are the foundation of organized, readable, and maintainable programs.

**In this notebook you will learn:**
- How to define and call functions
- Parameters, arguments, and default values
- Return values and scope
- Lambda functions
- Docstrings and recursion

## 1. Defining and Calling Functions

Use the `def` keyword to define a function. The function body must be indented.

```
def function_name(parameters):
    """Optional docstring."""
    # body
    return value
```

**Key rule:** `return` sends a value back to the caller. Without it, the function returns `None`.

````python
# Basic function definition and call
def greet(name):
    return f"Hello, {name}!"

message = greet("Alice")   # call the function, store the result
print(message)             # Hello, Alice!

# return vs print — critical difference
def add_print(a, b):
    print(a + b)           # prints but returns None

def add_return(a, b):
    return a + b           # returns the value — can be used further

result = add_return(3, 4)
print(result * 2)          # 14  — possible because we returned the value

result2 = add_print(3, 4)  # prints 7
print(result2)             # None — nothing was returned
````

## 2. Parameters and Arguments

| Type | Description | Example |
|---|---|---|
| Positional | Matched by order | `f(1, 2)` |
| Keyword | Matched by name | `f(b=2, a=1)` |
| Default | Used when argument is omitted | `def f(x, y=10)` |

> **Rule:** Parameters with defaults must come **after** parameters without defaults.

````python
# Positional arguments
def power(base, exponent):
    return base ** exponent

print(power(2, 10))          # 1024 — positional: base=2, exponent=10
print(power(exponent=3, base=5))  # 125 — keyword: order doesn't matter

# Default parameter values
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet("Alice"))              # Hello, Alice!  (uses default)
print(greet("Bob", "Good morning")) # Good morning, Bob!  (overrides default)

# Multiple parameters with mixed styles
def describe(name, age, city="Unknown"):
    return f"{name} is {age} years old from {city}."

print(describe("Ana", 28))
print(describe("Leo", 32, city="São Paulo"))
````

## 3. Return Values and Scope

**Multiple return values** — Python returns them as a tuple, which you can unpack.

**Scope** — variables defined inside a function are **local** and invisible outside it.

````python
# Multiple return values (returned as a tuple)
def min_max(numbers):
    smallest = numbers[0]
    largest = numbers[0]
    for n in numbers:
        if n < smallest:
            smallest = n
        if n > largest:
            largest = n
    return smallest, largest   # returns a tuple

data = [4, 7, 1, 9, 3]
low, high = min_max(data)      # tuple unpacking
print(f"Min: {low}, Max: {high}")  # Min: 1, Max: 9

# --- Scope ---
x = 100   # global variable

def show_scope():
    x = 42   # local variable — does NOT change the global x
    print(f"Inside function: x = {x}")

show_scope()
print(f"Outside function: x = {x}")   # still 100
````

## 4. Lambda Functions

A `lambda` is a small anonymous function defined in a single line.

```python
lambda parameters: expression
```

Use lambdas for **short, throwaway functions** — especially as arguments to `sorted()`, `map()`, or `filter()`.

````python
# Lambda vs equivalent def
square_def = lambda x: x ** 2
print(square_def(5))   # 25

# Most useful as inline arguments
students = [("Alice", 88), ("Bob", 72), ("Carol", 95)]

# Sort by grade (second element of each tuple)
sorted_students = sorted(students, key=lambda s: s[1], reverse=True)
print(sorted_students)  # [('Carol', 95), ('Alice', 88), ('Bob', 72)]

# Using lambda with a loop
numbers = [1, -3, 4, -2, 5, -1]
is_positive = lambda n: n > 0
positives = [n for n in numbers if is_positive(n)]
print(positives)   # [1, 4, 5]
````

## 5. Docstrings and `help()`

A **docstring** is a string literal placed as the first statement in a function. It documents what the function does.

```python
def my_function(x):
    """One-line summary.
    
    Args:
        x: description
    Returns:
        description
    """
```

Use `help(function_name)` to display the docstring.

````python
def celsius_to_fahrenheit(c):
    """Convert a temperature from Celsius to Fahrenheit.

    Args:
        c (float): Temperature in Celsius.

    Returns:
        float: Temperature in Fahrenheit.

    Examples:
        >>> celsius_to_fahrenheit(0)
        32.0
        >>> celsius_to_fahrenheit(100)
        212.0
    """
    return (c * 9 / 5) + 32

help(celsius_to_fahrenheit)   # displays the docstring
print(celsius_to_fahrenheit(37))  # 98.6

# --- Recursion example ---
def factorial(n):
    """Return the factorial of n (n!). Returns None for negative input."""
    if n < 0:
        return None
    if n == 0:          # base case
        return 1
    return n * factorial(n - 1)  # recursive call

for i in range(6):
    print(f"{i}! = {factorial(i)}")
````
