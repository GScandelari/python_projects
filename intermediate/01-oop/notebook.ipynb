# Object-Oriented Programming (OOP)

OOP organizes code around **objects** — bundles of data (attributes) and behavior (methods). Everything in Python is already an object: strings, lists, integers.

**In this notebook:**
- Classes and instances
- `__init__`, `self`, attributes
- Instance vs class attributes
- Encapsulation
- Inheritance and `super()`
- Dunder methods

## 1. Classes and Objects

A **class** is a blueprint. An **instance** is a concrete object built from it.

```
Class  →  Blueprint (Dog)
Object →  Specific instance (rex, fido)
```

````python
class Dog:
    species = "Canis lupus familiaris"   # class attribute — shared by all

    def __init__(self, name, breed, age):
        # instance attributes — unique per object
        self.name = name
        self.breed = breed
        self.age = age

    def bark(self):
        return f"{self.name} says: Woof!"

    def birthday(self):
        self.age += 1

    def __str__(self):
        return f"Dog({self.name}, {self.breed}, {self.age}yo)"


rex = Dog("Rex", "Labrador", 3)
fido = Dog("Fido", "Poodle", 5)

print(rex.bark())       # Rex says: Woof!
print(fido.bark())      # Fido says: Woof!
print(Dog.species)      # shared class attribute
rex.birthday()
print(rex)              # Dog(Rex, Labrador, 4yo)
print(fido.age)         # 5 — unchanged
````

## 2. Encapsulation

Hide internal details using naming conventions:

| Convention | Meaning |
|---|---|
| `name` | Public |
| `_name` | Protected (internal, avoid from outside) |
| `__name` | Private (name-mangled) |

Provide **methods** to interact with protected data safely.

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self._balance = balance        # protected — use methods to access

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive.")
        self._balance += amount

    def withdraw(self, amount):
        if amount <= 0 or amount > self._balance:
            raise ValueError("Invalid withdrawal amount.")
        self._balance -= amount

    def get_balance(self):
        return self._balance

    def __str__(self):
        return f"Account[{self.owner}]: R$ {self._balance:.2f}"


acc = BankAccount("Alice", 1000)
acc.deposit(500)
acc.withdraw(200)
print(acc.get_balance())   # 1300
print(acc)                 # Account[Alice]: R$ 1300.00
````

## 3. Inheritance

A **child class** inherits all attributes and methods from the **parent** and can extend or override them.

Use `super()` to call the parent's version of a method.

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f"{self.name} makes a sound."

    def __str__(self):
        return f"{type(self).__name__}({self.name})"


class Dog(Animal):
    def speak(self):
        return f"{self.name} says: Woof!"


class Cat(Animal):
    def speak(self):
        return f"{self.name} says: Meow!"


class GuideDog(Dog):
    def __init__(self, name, owner):
        super().__init__(name)          # call Dog → Animal __init__
        self.owner = owner

    def speak(self):
        base = super().speak()          # call Dog.speak()
        return f"{base} (Guide dog for {self.owner})"


animals = [Dog("Rex"), Cat("Whiskers"), GuideDog("Buddy", "Alice")]
for a in animals:
    print(a)             # uses __str__ from Animal
    print(a.speak())     # uses each class's own speak()
````

## 4. Method Overriding and Polymorphism

**Polymorphism** means the same method call behaves differently depending on the object's type. This lets you write generic code that works with any subclass.

````python
class Shape:
    def area(self):
        return 0

    def describe(self):
        return f"I am a {type(self).__name__} with area {self.area():.2f}"


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2

    def __str__(self):
        return f"Circle(radius={self.radius})"


class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h

    def area(self):
        return self.w * self.h

    def __str__(self):
        return f"Rectangle({self.w}x{self.h})"


shapes = [Circle(5), Rectangle(4, 6), Circle(3)]

# Generic loop — works for any Shape subclass
for s in shapes:
    print(s.describe())

# Total area — polymorphism makes this trivial
total = sum(s.area() for s in shapes)
print(f"\nTotal area: {total:.2f}")
````

## 5. Dunder Methods

Dunder (magic) methods let your classes integrate with Python's built-in syntax.

| Method | Triggered by |
|---|---|
| `__str__` | `print(obj)`, `str(obj)` |
| `__repr__` | REPL display, `repr(obj)` |
| `__len__` | `len(obj)` |
| `__eq__` | `obj1 == obj2` |
| `__add__` | `obj1 + obj2` |
| `__lt__` | `obj1 < obj2` |

````python
class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"Vector2D({self.x}, {self.y})"

    def __repr__(self):
        return f"Vector2D(x={self.x}, y={self.y})"

    def __add__(self, other):
        return Vector2D(self.x + other.x, self.y + other.y)

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    def magnitude(self):
        return round((self.x**2 + self.y**2) ** 0.5, 4)


v1 = Vector2D(3, 4)
v2 = Vector2D(1, 2)

print(v1 + v2)               # Vector2D(4, 6)
print(v1.magnitude())        # 5.0
print(v1 == Vector2D(3, 4))  # True
print(v1 == v2)              # False
````
